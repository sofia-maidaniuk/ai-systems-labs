## 1. Завантаження моделей

[**SmolLM3**](https://huggingface.co/blog/smollm3) — відкрита модель від Hugging Face на 3 млрд параметрів. Вона має дві версії:
- **Base** — лише після попереднього тренування: продовжує текст, але не веде діалог;
- **Instruct** (`SmolLM3-3B`) — донавчена виконувати інструкції й вести діалог.

```python
base_model_name = "HuggingFaceTB/SmolLM3-3B-Base"
instruct_model_name = "HuggingFaceTB/SmolLM3-3B"

base_tokenizer = AutoTokenizer.from_pretrained(base_model_name)
instruct_tokenizer = AutoTokenizer.from_pretrained(instruct_model_name)

# Ваги SmolLM3 збережено в bfloat16, але T4 не підтримує його апаратно,
# тому завантажуємо у float16.
base_model = AutoModelForCausalLM.from_pretrained(
    base_model_name, dtype=torch.float16, device_map="auto"
)
instruct_model = AutoModelForCausalLM.from_pretrained(
    instruct_model_name, dtype=torch.float16, device_map="auto"
)
print("Моделі завантажено")
```

## 2. Як шаблон розмови форматує діалог

Діалог задається списком повідомлень із ролями:
- `system` — інструкції для моделі (роль, стиль, обмеження);
- `user` — повідомлення користувача;
- `assistant` — відповіді моделі.

Метод `apply_chat_template` перетворює цей список на рядок зі спеціальними токенами.

Параметр `add_generation_prompt`:
- `False` — для передачі в модель завершених діалогів, тобто для **тренувальних даних**;
- `True` — додає в кінець заголовок відповіді асистента, щоб модель почала відповідати; використовується для **інференсу**.

Для SmolLM3 у системне повідомлення можна додати прапорець `/think` або `/no_think`,
який вмикає чи вимикає режим міркування.

```python
# Різні типи діалогів для порівняння
conversations = {
    "simple_qa": [
        {"role": "system", "content": "/no_think"},
        {"role": "user", "content": "What is machine learning?"},
    ],
    "with_system": [
        {
            "role": "system",
            "content": "You are a helpful AI assistant specialized in explaining technical concepts clearly. /no_think",
        },
        {"role": "user", "content": "What is machine learning?"},
    ],
    "multi_turn": [
        {"role": "system", "content": "You are a math tutor. /no_think"},
        {"role": "user", "content": "What is calculus?"},
        {
            "role": "assistant",
            "content": "Calculus is a branch of mathematics that deals with rates of change and accumulation of quantities.",
        },
        {"role": "user", "content": "Can you give me a simple example?"},
    ],
    "reasoning_task": [
        {"role": "system", "content": "/think"},
        {
            "role": "user",
            "content": "Solve step by step: If a train travels 120 miles in 2 hours, what is its average speed?",
        },
    ],
}

for conv_type, messages in conversations.items():
    print(f"--- {conv_type.upper()} ---")

    # Завершений діалог: так виглядають тренувальні дані
    formatted_complete = instruct_tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=False
    )
    # Діалог для інференсу: у кінці додано заголовок відповіді асистента
    formatted_prompt = instruct_tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )

    print("Завершений діалог (add_generation_prompt=False):")
    print(formatted_complete)
    print("\nДля інференсу (add_generation_prompt=True):")
    print(formatted_prompt)
    print("\n" + "=" * 50 + "\n")
```

**Що варто помітити:**
- Шаблон SmolLM3 сам додає до системного повідомлення блок метаданих (дата відсічення знань, поточна дата, режим міркування)
  і стандартну інструкцію, якщо власної немає. Модель бачить більше, ніж ми написали.
- Кожне повідомлення обгорнуте токенами `<|im_start|>роль` … `<|im_end|>` (формат ChatML).
- З `add_generation_prompt=True` рядок закінчується на `<|im_start|>assistant`: з цього місця модель починає генерувати.

## 3. Base проти Instruct

Ставимо обом моделям одне питання. Base-модель отримує сирий текст без шаблону,
instruct-модель — діалог, відформатований шаблоном розмови.

```python
def generate(model, tokenizer, text, max_new_tokens=150, temperature=0.7):
    """Генерує продовження тексту й повертає лише нові токени, без промпту."""
    inputs = tokenizer(text, return_tensors="pt").to(model.device)
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            temperature=temperature,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id,
        )
    # Відкидаємо токени промпту: декодуємо лише те, що згенерувала модель
    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]
    return tokenizer.decode(new_tokens, skip_special_tokens=True)
```

```python
test_prompt = "Explain me, what is quantum computing?"

# Base-модель: сирий текст без шаблону
print("🤖 BASE-МОДЕЛЬ:")
print("-" * 50)
print(generate(base_model, base_tokenizer, test_prompt))

# Instruct-модель: діалог, відформатований шаблоном розмови
instruct_messages = [
    {"role": "system", "content": "/no_think"},
    {"role": "user", "content": test_prompt},
]
instruct_formatted = instruct_tokenizer.apply_chat_template(
    instruct_messages, tokenize=False, add_generation_prompt=True
)
print("\n" + "=" * 50)
print("💬 INSTRUCT-МОДЕЛЬ:")
print("-" * 50)
print(generate(instruct_model, instruct_tokenizer, instruct_formatted))
```

## 4. Режими міркування `/think` і `/no_think`

SmolLM3 — гібридна модель: у режимі `/think` вона спочатку генерує міркування в блоці `<think>…</think>`,
а потім відповідь; у режимі `/no_think` блок міркувань порожній.

```python
reasoning_prompts = [
    "What is 15 × 24? Show your work.",
    "A recipe calls for 2 cups of flour for 12 cookies. How much flour is needed for 30 cookies?",
    "If I have $50 and spend $18.75 on lunch and $12.30 on a book, how much money do I have left?",
]

for thinking_flag in ["/no_think", "/think"]:
    print(f"===== Режим: {thinking_flag} =====\n")
    for i, prompt in enumerate(reasoning_prompts, 1):
        messages = [
            {"role": "system", "content": thinking_flag},
            {"role": "user", "content": prompt},
        ]
        formatted = instruct_tokenizer.apply_chat_template(
            messages, tokenize=False, add_generation_prompt=True
        )
        # Нижча температура робить відповіді на арифметичні задачі стабільнішими
        answer = generate(instruct_model, instruct_tokenizer, formatted, max_new_tokens=300, temperature=0.3)
        print(f"Задача {i}: {prompt}")
        print(f"Відповідь: {answer}")
        print("\n" + "-" * 50 + "\n")
```

## 5. У кожної моделі свій шаблон

Ролі `system`, `user`, `assistant` — це домовленість на рівні формату `messages`.
У токенах кожна модель записує їх по-своєму. Порівняємо шаблони кількох моделей.

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Hi!"},
    {"role": "assistant", "content": "Hello! How can I help?"},
]

for name in ["HuggingFaceTB/SmolLM3-3B", "Qwen/Qwen3-0.6B", "microsoft/Phi-3.5-mini-instruct"]:
    tok = AutoTokenizer.from_pretrained(name)
    print(f"===== {name} =====")
    print(tok.apply_chat_template(messages, tokenize=False))
    print()
```

Прапорці режиму міркування також залежать від моделі. У Qwen3, наприклад,
режим вмикається параметром `enable_thinking` у `apply_chat_template`.

```python
qwen_tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen3-0.6B")
question = [{"role": "user", "content": "What is 2 + 2?"}]

for enable_thinking in [True, False]:
    print(f"===== enable_thinking={enable_thinking} =====")
    print(qwen_tokenizer.apply_chat_template(
        question, tokenize=False, add_generation_prompt=True, enable_thinking=enable_thinking
    ))
```

## 6. Підготовка датасету для SFT

Для SFT дані зберігають у форматі `messages` — тобто у списку повідомлень із ролями.
Бібліотека TRL (`SFTTrainer`) сама застосовує до нього шаблон розмови під час тренування.

### SmolTalk2

**SmolTalk2** — датасет, на якому донавчали SmolLM3. Він складається з багатьох частин (splits) з різних джерел.
Список частин можна отримати без завантаження самих даних.

```python
splits = get_dataset_split_names("HuggingFaceTB/smoltalk2", "SFT")
print(f"Кількість частин: {len(splits)}")
for s in splits:
    print(" -", s)
```

```python
# Завантажуємо лише файл однієї частини через data_files: повний датасет займає десятки гігабайт,
# а load_dataset(..., split=...) спершу перебирає файли всіх частин і працює дуже довго
split = "smoltalk_everyday_convs_reasoning_Qwen3_32B_think"
everyday = load_dataset("HuggingFaceTB/smoltalk2", data_files=f"SFT/{split}-*.parquet", split="train")
print(everyday)
print()
for message in everyday[0]["messages"]:
    print(f"[{message['role']}] {message['content'][:200]}")
```

Ця частина зібрана для режиму `/think`, тому відповіді асистента містять блок міркувань `<think>…</think>`.

Як видно з назви частини, відповіді в ній згенерувала модель Qwen3-32B. Такі **синтетичні дані** від сильнішої моделі —
поширений спосіб отримати тисячі прикладів для SFT без ручного написання відповідей. Помилки моделі-генератора при цьому переходять у дані,
тому синтетичні дані фільтрують.

### Перетворення власного датасету у формат `messages`

Більшість датасетів має інший формат, наприклад колонки «питання» і «відповідь».
Перетворимо датасет задач з математики **GSM8K** у формат `messages`.

```python
gsm8k = load_dataset("openai/gsm8k", "main", split="train[:100]")  # невелика вибірка для демо
print("Оригінальний приклад:", gsm8k[0])


def to_messages(examples):
    """Перетворює пари «питання — відповідь» на діалоги з ролями."""
    processed = []
    for question, answer in zip(examples["question"], examples["answer"]):
        processed.append([
            {"role": "system", "content": "You are a math tutor. Solve problems step by step."},
            {"role": "user", "content": question},
            {"role": "assistant", "content": answer},
        ])
    return {"messages": processed}


gsm8k_messages = gsm8k.map(to_messages, batched=True, remove_columns=gsm8k.column_names)
print("\nУ форматі messages:", gsm8k_messages[0])
```

### Що побачить модель під час тренування

Застосуємо шаблон розмови до підготовленого прикладу. Саме цей рядок модель отримає під час SFT.

```python
text = instruct_tokenizer.apply_chat_template(
    gsm8k_messages[0]["messages"],
    tokenize=False,
    add_generation_prompt=False,  # діалог завершений: відповідь асистента вже є
)
print(text)
print(f"Довжина в токенах: {len(instruct_tokenizer(text)['input_ids'])}")
```

**Що варто помітити:** шаблон знову додав метадані й режим `/think`, хоча в даних їх немає,
а відповідь асистента не містить блоку міркувань. Якщо донавчати на таких прикладах, модель вчитиметься
відповідати без міркувань навіть у режимі `/think`. Тому формат даних і шаблон мають узгоджуватися з тим,
як модель використовуватиметься після донавчання.

## 12. Для чого змінювати параметр add_generation_prompt в методі apply_chat_template() при використанні моделі? Як по згенерованим розмовам можна дізнатися значення цього параметру, яке було задано при виклику метода?

Параметр `add_generation_prompt` у методі `apply_chat_template()` визначає, чи потрібно додавати в кінці діалогу початок нового повідомлення асистента

Якщо встановлено `False`, шаблон лише форматує наявні повідомлення без додавання початку нової відповіді. Такий варіант використовують переважно для підготовки навчальних даних, де відповіді вже записані

Якщо встановлено `True`, шаблон додає в кінці спеціальний запис `<|im_start|>assistant`, який вказує моделі, що потрібно почати генерувати відповідь. Цей варіант використовують під час спілкування з моделлю

Визначити значення параметра можна за результатом форматування діалогу. У демо при `True` наприкінці з'являється `<|im_start|>assistant`, а в режимі `/no_think` також порожній блок `<think></think>`. При `False` ці додаткові елементи не додаються

## 13. Чим відрізняються ролі system від assistant в шаблонах розмов?

system використовується для задання інструкцій моделі. Через цю роль визначають, як модель повинна поводитися, у якому стилі відповідати та яких правил дотримуватися. Наприклад, у демо через system моделі задають роль викладача математики

assistant позначає відповіді самої моделі на повідомлення користувача. Під час навчання в цій ролі записують правильні приклади відповідей, на яких модель навчається

System задає правила поведінки моделі, а assistant містить відповіді, які модель формує відповідно до цих правил

## 14. Прокоментуйте вивід даних в секції 2 «Як шаблон розмови форматує діалог». Які секції було додано шаблонізатором розмов і для чого?

У секції 2 видно, що метод `apply_chat_template()` не просто об'єднує повідомлення в один текст, а й додає службову інформацію, яка допомагає моделі правильно обробляти діалог

Шаблон додав секцію `Metadata`, яка містить дату завершення знань моделі, поточну дату та режим роботи (`/think` або `/no_think`). Також було додано секцію `Custom Instructions`, у якій записуються інструкції для моделі, наприклад, яку роль вона повинна виконувати та як відповідати користувачу. У режимі `/think` шаблон також додає детальніші інструкції щодо міркування перед відповіддю

Крім цього, шаблон використовує спеціальні токени `<|im_start|>` і `<|im_end|>`, щоб позначати початок і кінець повідомлень та розрізняти ролі `system`, `user` і `assistant`

У демо також видно різницю між значеннями `add_generation_prompt`. При `True` наприкінці додається початок нового повідомлення асистента, щоб модель могла згенерувати відповідь. При `False` цей початок не додається, оскільки шаблон лише форматує наявний діалог

Додані секції та спеціальні токени допомагають моделі розуміти структуру розмови, дотримуватися інструкцій і правильно формувати відповіді

## 15. Прокоментуйте різницю між Base і Instruct, проглянувши вивід в секції 3.

У секції 3 порівнюються дві версії моделі SmolLM3: Base та Instruct. Обидві отримали однакове запитання про квантові обчислення, але їхні відповіді відрізнялися

Base - це базова модель, яка навчена передбачати наступні слова в тексті, але не донавчена спеціально для спілкування з користувачем. У демо вона почала відповідь із додаткових запитань, а потім перейшла до пояснення квантових обчислень. Її відповідь була менш послідовною та залишилася незавершеною

Instruct - це модель, яку додатково навчили виконувати інструкції та відповідати на запитання користувачів. У демо вона одразу почала пояснювати, що таке квантові обчислення, розповіла про їхні основні принципи та структурувала відповідь

Тому основна різниця полягає в тому, що Base переважно продовжує отриманий текст, а Instruct краще розуміє запитання та формує відповідь відповідно до інструкцій

## 16. Чому в режимі /think при генерації було передано параметр max_new_tokens=300?

У режимі /think модель спочатку формує пояснення своїх міркувань, а потім надає остаточну відповідь. Тому їй потрібно більше токенів, ніж у режимі /no_think

Параметр max_new_tokens=300 встановлює максимальну кількість токенів, які модель може згенерувати для однієї відповіді, включно з міркуваннями. Це дозволяє обмежити довжину відповіді та час її генерації

У демо видно, що 300 токенів виявилося недостатньо, оскільки модель витратила їх на міркування та не встигла сформувати остаточну відповідь

## 17. Чи впоралася модель з математичними задачами? Який би режим генерації Ви обрали для себе, якщо б Вам потрібна була допомога?

У секції 4 модель перевіряли на трьох математичних задачах у режимах `/no_think` та `/think`

У режимі `/no_think` модель правильно розв'язала всі три задачі: обчислила 15 × 24 = 360, визначила, що для 30 печив потрібно 5 чашок борошна, та підрахувала залишок грошей 18,95 долара. Відповіді були повними та містили зрозумілі пояснення обчислень

У режимі `/think` модель також виконувала правильні обчислення, але витрачала більше токенів на міркування та перевірку результатів. Через обмеження `max_new_tokens=300` вона не встигала завершити відповіді

Для себе я б обрала режим `/no_think` для простих математичних задач, оскільки він швидше надає правильні та зрозумілі відповіді. Для складніших задач можна використовувати `/think`, але зі збільшеним обмеженням кількості токенів, щоб модель могла завершити розв'язання

## 18. Які відмінності в шаблонах розмов Ви помітили в секції 5?

У секції 5 порівнюються шаблони розмов трьох моделей: SmolLM3, Qwen3 та Phi-3.5. Для всіх моделей використали однаковий діалог із ролями system, user та assistant, але після форматування його структура відрізнялася

У SmolLM3 шаблон додав секцію Custom Instructions із системними інструкціями. Для позначення повідомлень використовуються спеціальні токени <|im_start|> та <|im_end|>

У Qwen3 також використовуються <|im_start|> та <|im_end|>, але системне повідомлення записується без додаткової секції Custom Instructions. Плюс перед відповіддю асистента з'явився порожній блок <think></think>, призначений для міркувань моделі

У Phi-3.5 використовується інший формат, ролі позначаються токенами <|system|>, <|user|> та <|assistant|>, повідомлення завершуються через <|end|>, а наприкінці діалогу додається <|endoftext|>

Хоча всі три моделі працюють з однаковими ролями, кожна має власні правила форматування розмови. Тому під час роботи з моделлю та її донавчання важливо використовувати саме той шаблон, для якого вона була підготовлена

## 19. Для чого нам потрібно знати про шаблони розмов, якщо ми хочемо донавчити модель за допомогою SFT?

Під час донавчання моделі методом SFT (Supervised Fine-Tuning) важливо знати її шаблон розмови, оскільки саме він визначає, як навчальні дані перетворюються на текст, який отримує модель

Шаблон правильно розділяє повідомлення за ролями system, user та assistant, додає необхідні спеціальні токени й показує моделі, де закінчується запитання та починається відповідь

Якщо використовувати неправильний шаблон, модель може плутати ролі, неправильно визначати межі повідомлень або генерувати зайві службові токени. Це може погіршити результати донавчання

У демо також показано, що шаблони різних моделей відрізняються, тому під час SFT потрібно використовувати формат, який відповідає конкретній моделі. Важливо також узгоджувати формат навчальних даних із режимом роботи моделі, наприклад /think або /no_think

Знання шаблонів розмов потрібне для правильної підготовки навчальних даних, щоб модель навчилася відповідати у потрібному форматі та коректно працювала після донавчання

## 20. З яких даних було побудовано датасет HuggingFaceTB/smoltalk2?

Датасет HuggingFaceTB/smoltalk2 створений компанією Hugging Face для донавчання моделі SmolLM3. Він об'єднує дані з різних джерел і містить приклади діалогів, виконання інструкцій, математичних задач, програмування та логічних міркувань

Датасет складається з трьох основних частин: Mid - для додаткового навчання моделі, SFT - для навчання відповідати на інструкції користувачів, та Preference - для покращення якості відповідей на основі порівняння різних варіантів

У демо розглядалася частина SFT, яка складається з 25 наборів даних із різних джерел, зокрема SmolTalk, OpenThoughts3, OpenHermes, LongAlign та інших. Частина прикладів була згенерована за допомогою потужнішої моделі Qwen3-32B

Навчальні приклади поділені на два типи: think - із поясненнями міркувань моделі, та no_think - без них. Це дозволяє навчати модель працювати в обох режимах

У ноутбуці було завантажено одну з частин датасету smoltalk_everyday_convs_reasoning_Qwen3_32B_think, яка містила 2057 прикладів діалогів із міркуваннями асистента

## 21. Які модифікації даних відбуваються в секції "Перетворення власного датасету у формат messages"?

У секції «Перетворення власного датасету у формат messages» показано, як підготувати звичайний датасет для донавчання моделі методом SFT

Для прикладу використано датасет GSM8K, який містить математичні задачі та їхні розв'язання. Спочатку з нього завантажили 100 прикладів, де кожен запис складався з двох полів: question та answer

За допомогою функції to_messages() ці дані перетворили на діалог із трьома ролями: system, user та assistant. До кожного прикладу додали повідомлення system з інструкцією бути викладачем математики та розв'язувати задачі покроково. Саме запитання перенесли в повідомлення user, а готову відповідь у assistant

Перетворення виконали за допомогою методу map(), після чого початкові колонки question та answer видалили, залишивши нову колонку messages

Зміст математичних задач і відповідей не змінювався, але їхню структуру адаптували до формату діалогу. Це дозволяє надалі застосувати шаблон розмови та використовувати підготовлені дані для SFT
